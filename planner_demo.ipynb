```python
from caldera_helper import request, save_json

cheese_bagel_id = 'cheese_bagel_id'
cheese_bagel_operation_id = 'cheese_bagel_operation_id'
```

```python
obj = request(
    'POST',
    'objectives',
    {
        "name": "stage files objective",
        "description": "cheese description",
        "goals": [
            {
                "target": "host.dir.staged",
                "value": "N/A",
                "operator": "*",
                "count": 1
            }
        ]
    }
)
save_json(obj, 'objective')
obj
```

```python
planners = request('GET', 'planners')
print(planners[-1])
```

```python
# get everything bagel
adversary = request('GET', 'adversaries/785baa02-df5d-450a-ab3a-1a863f22b4b0')

# change objective and give new name/id
adversary['objective'] = obj['id']
adversary['name'] = 'cheese bagel'
adversary['adversary_id'] = cheese_bagel_id
cheese_bagel_adversary = request('POST', 'adversaries', adversary)
cheese_bagel_adversary
```

```python
# create operation with cheese bagel adversary
operation = request(
    'POST',
    'operations',
    {
        "id": cheese_bagel_operation_id,
        "name": "cheese bagel operation",
        "adversary": {"adversary_id": cheese_bagel_id},
        "planner": {"id": planners[-1]['id']}, # look ahead planner
        "source": {"id": "basic"}
    }
)
save_json(operation, 'operation')
```
